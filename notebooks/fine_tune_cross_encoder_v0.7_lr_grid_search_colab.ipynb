# Cross-Encoder v0.7 — Phase 1: LR Grid Search

## Hypothesis
Thay `ms-marco-MiniLM-L-12-v2` (384-dim) bằng `mmarco-MiniLMv2-L12-H768-v1` (768-dim).  
Cùng MS MARCO pre-training → không bị lỗi như DistilRoBERTa.  
Model lớn hơn có thể học representation phân biệt adjacent classes tốt hơn (đặc biệt `weak_match` chỉ đạt 52.3% ở v0.6).

## Tại sao cần LR Grid Search?
Model 768-dim lớn hơn 384-dim → gradient flow khác → LR optimal có thể thấp hơn 5e-5.  
Test 3 LRs với seed=42 cố định để isolate hiệu ứng LR, không lẫn seed variance.

| LR | Lý do |
|---|---|
| 2e-5 | Conservative — chuẩn BERT fine-tune |
| 3e-5 | Middle ground — tốt với 384-dim ở v0.6 |
| 5e-5 | Aggressive — optimal v0.6, có thể quá cao với 768-dim |

## v0.6 Baseline
| | Best single | Ensemble |
|---|---|---|
| LabelAcc | 64.95% | **65.80%** |
| MAE | 9.27 | 9.11 |

## Output
- Report: `artifacts/reports/fine_tune_cross_encoder_v0.7_lr_grid_search_report.json`
- Dùng best LR cho Phase 2 (Ensemble)

```python
# === CELL 2: Load Data & Helpers ===
import json
import numpy as np
import random
from scipy.stats import spearmanr
from sentence_transformers import CrossEncoder, InputExample
from torch.utils.data import DataLoader

BASE_MODEL = 'cross-encoder/mmarco-MiniLMv2-L12-H768-v1'
DATA_DIR   = 'datasets/versions/v0.5/cross_encoder'

V06_BEST_SINGLE  = 0.6495  # lr5e-05-15ep
V06_ENSEMBLE_ACC = 0.6580


class CECorrelationEvaluator:
    def __init__(self, sentence_pairs, labels_0_1):
        self.sentence_pairs = sentence_pairs
        self.labels_0_1     = labels_0_1

    @classmethod
    def from_input_examples(cls, examples):
        return cls([ex.texts for ex in examples], [ex.label for ex in examples])

    def __call__(self, model, output_path=None, epoch=-1, steps=-1):
        preds = model.predict(self.sentence_pairs, batch_size=32, show_progress_bar=False)
        corr, _ = spearmanr(preds, self.labels_0_1)
        return float(corr) if not np.isnan(corr) else 0.0


def load_jsonl(path):
    with open(path, 'r', encoding='utf-8') as f:
        return [json.loads(l) for l in f if l.strip()]


def to_examples(records):
    return [InputExample(texts=[r['cv_text'], r['jd_text']], label=r['label']) for r in records]


def compute_metrics(model, examples):
    preds      = model.predict([ex.texts for ex in examples], batch_size=32, show_progress_bar=False)
    preds_100  = np.asarray(preds) * 100
    labels_100 = np.array([ex.label * 100 for ex in examples])
    return {
        'LabelAcc': float(np.mean(np.abs(preds_100 - labels_100) <= 10)),
        'MAE':      float(np.mean(np.abs(preds_100 - labels_100))),
        'RMSE':     float(np.sqrt(np.mean((preds_100 - labels_100) ** 2))),
    }


train_examples = to_examples(load_jsonl(f'{DATA_DIR}/cross_encoder_train.jsonl'))
val_examples   = to_examples(load_jsonl(f'{DATA_DIR}/cross_encoder_validation.jsonl'))
test_examples  = to_examples(load_jsonl(f'{DATA_DIR}/cross_encoder_test.jsonl'))

print(f'Data: {len(train_examples)} train | {len(val_examples)} val | {len(test_examples)} test')
print(f'Model: {BASE_MODEL}')
```

```python
# === CELL 3: LR Grid Search ===
from google.colab import drive
drive.mount('/content/drive')

LR_CANDIDATES = [2e-5, 3e-5, 5e-5]
SEED          = 42
EPOCHS        = 15
BATCH_SIZE    = 16

total_steps  = (len(train_examples) // BATCH_SIZE + 1) * EPOCHS
warmup_steps = int(total_steps * 0.1)

print(f'LR Grid Search — {BASE_MODEL}')
print(f'Seed={SEED} | BS={BATCH_SIZE} | Epochs={EPOCHS} | Warmup={warmup_steps} steps')
print(f'Testing: {LR_CANDIDATES}\n')

results = []

for lr in LR_CANDIDATES:
    torch.manual_seed(SEED)
    np.random.seed(SEED)
    random.seed(SEED)

    run_name   = f'v0.7-lr{lr:.0e}-seed{SEED}'
    output_dir = f'artifacts/models/cross-encoder-cv-jd-{run_name}'

    print(f'\n--- LR = {lr:.0e} ---')
    model      = CrossEncoder(BASE_MODEL, num_labels=1, default_activation_function=torch.nn.Sigmoid())
    evaluator  = CECorrelationEvaluator.from_input_examples(val_examples)
    dataloader = DataLoader(train_examples, shuffle=True, batch_size=BATCH_SIZE)

    model.fit(
        train_dataloader=dataloader,
        evaluator=evaluator,
        epochs=EPOCHS,
        loss_fct=torch.nn.MSELoss(),
        optimizer_params={'lr': lr},
        warmup_steps=warmup_steps,
        output_path=output_dir,
        save_best_model=True,
        use_amp=True,
        max_grad_norm=1.0,
        show_progress_bar=True,
    )

    best_model = CrossEncoder(output_dir)
    val_m      = compute_metrics(best_model, val_examples)
    test_m     = compute_metrics(best_model, test_examples)

    vs_v06 = (test_m['LabelAcc'] - V06_BEST_SINGLE) * 100
    print(f'  Val  LabelAcc: {val_m["LabelAcc"]*100:.2f}%  | MAE: {val_m["MAE"]:.2f}')
    print(f'  Test LabelAcc: {test_m["LabelAcc"]*100:.2f}%  | MAE: {test_m["MAE"]:.2f}  ({vs_v06:+.2f}pp vs v0.6 best single 64.95%)')

    results.append({
        'learning_rate': lr,
        'run': run_name,
        'val':  {'LabelAcc': val_m['LabelAcc'], 'MAE': val_m['MAE'], 'RMSE': val_m['RMSE']},
        'test': {'LabelAcc': test_m['LabelAcc'], 'MAE': test_m['MAE'], 'RMSE': test_m['RMSE']},
    })

    # Upload model to Drive
    try:
        import shutil
        dest = f'/content/drive/MyDrive/ai-recruiter/models/{run_name}'
        if os.path.exists(dest): shutil.rmtree(dest)
        shutil.copytree(output_dir, dest)
        print(f'  Uploaded to Drive: {run_name}')
    except Exception as e:
        print(f'  Drive upload skipped: {e}')

# Summary
best = max(results, key=lambda r: r['test']['LabelAcc'])

print(f'\n{"="*58}')
print(f'  Phase 1 Summary')
print(f'{"="*58}')
print(f'{"LR":<10} {"Val Acc":>10} {"Test Acc":>10} {"vs v0.6":>10}')
print(f'{"-"*42}')
for r in results:
    vs = (r['test']['LabelAcc'] - V06_BEST_SINGLE) * 100
    mark = ' ← best' if r['learning_rate'] == best['learning_rate'] else ''
    print(f'{r["learning_rate"]:<10.0e} {r["val"]["LabelAcc"]*100:>9.2f}%  {r["test"]["LabelAcc"]*100:>9.2f}%  {vs:>+8.2f}pp{mark}')

print(f'\n  Best LR: {best["learning_rate"]:.0e}  →  Test {best["test"]["LabelAcc"]*100:.2f}%')
print(f'  → Dùng LR={best["learning_rate"]:.0e} cho Phase 2 Ensemble')

# Save report
os.makedirs('artifacts/reports', exist_ok=True)
report_path = 'artifacts/reports/fine_tune_cross_encoder_v0.7_lr_grid_search_report.json'
with open(report_path, 'w') as f:
    json.dump({
        'experiment':   'Phase 1: LR Grid Search',
        'base_model':   BASE_MODEL,
        'dataset':      'v0.5',
        'seed':         SEED,
        'epochs':       EPOCHS,
        'batch_size':   BATCH_SIZE,
        'results':      results,
        'best': {
            'learning_rate':  best['learning_rate'],
            'test_labelacc':  best['test']['LabelAcc'],
            'improvement_pp': (best['test']['LabelAcc'] - V06_BEST_SINGLE) * 100,
        },
    }, f, indent=2)
print(f'\n  Report: {report_path}')

try:
    import shutil
    shutil.copy(report_path, '/content/drive/MyDrive/ai-recruiter/reports/fine_tune_cross_encoder_v0.7_lr_grid_search_report.json')
    print('  Report uploaded to Drive')
except Exception as e:
    print(f'  Drive upload skipped: {e}')
```
